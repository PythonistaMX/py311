## Personalización de las figuras mediante métodos de *plt*.

Los objetos *figure* y *axes* de *Matplotlib* ofrecen una gran variedad de métodos para personalizar las figuras y los gráficos, como por ejemplo:
* [`plt.title()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.title.html) para establecer el título de la figura.
* [`plt.xlabel()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlabel.html) para establecer la etiqueta del eje $x$.
* [`plt.ylabel()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.ylabel.html) para establecer la etiqueta del eje $y$.
* [`plt.xlim()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlim.html) para establecer los límites del eje $x$.
* [`plt.ylim()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.ylim.html) para establecer los límites del eje $y$.
* [`plt.legend()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html) para mostrar la leyenda del gráfico.
* [`plt.grid()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.grid.html) para mostrar la cuadrícula del gráfico.
* [`plt.axis()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axis.html) para establecer los límites de los ejes y la relación de aspecto del gráfico.
* [`plt.tight_layout()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.tight_layout.html) para ajustar automáticamente el diseño de la figura para evitar solapamientos entre los elementos del gráfico.
* [`plt.savefig()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.savefig.html) para guardar la figura en un archivo.
* [`plt.show()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.show.html) para mostrar la figura en pantalla.
* [`plt.close()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.close.html) para cerrar la figura.

La lista de métodos para personalizar las figuras y los gráficos en *Matplotlib* es muy extensa, por lo que se recomienda consultar la documentación oficial de *Matplotlib* para obtener más información sobre los métodos disponibles.

https://matplotlib.org/stable/api/pyplot_api.html

### Título del gráfico.

El método `plt.title()` permite establecer el título de la figura.

```plt.title(<título>, fontsize=<tamaño>, fontweight=<peso>, color=<color>, **kwargs)```

Donde:
* `<título>` es una cadena de texto que representa el título de la figura.
* `<tamaño>` es el tamaño de la fuente utilizada para el título, el cual se
puede especificar como un número entero o un número de punto flotante. El valor por defecto es `12`.
* `<peso>` es el peso de la fuente utilizada para el título, el cual se puede especificar utilizando los siguientes valores:
    * `'light'` para un peso de fuente ligero.
    * `'normal'` para un peso de fuente normal.
    * `'bold'` para un peso de fuente negrita.
    * `'heavy'` para un peso de fuente pesado.
    * `'black'` para un peso de fuente negro.
* `<color>` es el color de la fuente utilizada para el título, el cual se puede especificar utilizando nombres de colores reconocidos por Matplotlib o códigos de color en formato hexadecimal.

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.title.html

**Ejemplo:**

* La siguiente celda trazará un gráfico que muestre el título `Pythonista`.

````python
plt.figure(dpi=80)
plt.title("Pythonista", fontweight="bold") # Título del gráfico
plt.plot(x, y, label='función 1')
plt.plot(x, z, label='función 2')
plt.legend(loc="lower right")
````

### Etiquetas de ejes.

Los métodos `plt.xlabel()` y `plt.ylabel()` permiten definir las etiquetas de los ejes $x$ e $y$ respectivamente.

```
plt.xlabel(<etiqueta>, **kwargs)
```
```
plt.ylabel(<etiqueta>, **kwargs)
```
Donde:
* `<etiqueta>` es una cadena de texto que representa la etiqueta del eje, la cual se mostrará en el gráfico. El valor por defecto es `''`, lo que significa que no se mostrará ninguna etiqueta en el eje.
Los métodos `plt.xlabel()` y `plt.ylabel()` también aceptan argumentos adicionales para personalizar la apariencia de las etiquetas, como por ejemplo:
* `fontsize` para especificar el tamaño de la fuente utilizada en las etiquetas.
* `color` para especificar el color de las etiquetas.
* `fontweight` para especificar el peso de la fuente utilizada en las etiquetas (normal, bold, etc.).
* `fontfamily` para especificar la familia de la fuente utilizada en las etiquetas (serif, sans-serif, etc.).
* `rotation` para especificar la rotación de las etiquetas en grados.

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlabel.html

**Ejemplos:**

* La siguiente celda trazará una imagen con las etiquetas:
 * ```Eje de las x``` para el eje $x$.
 * ```f(x)``` para el eje $y$.

````python
plt.figure(dpi=80)
plt.title("Pythonista")
plt.plot(x, y, label='función 1')
plt.plot(x, z, label='función 2')
plt.legend(loc="lower right")
plt.xlabel('Eje de las x', color='blue', fontsize=12, fontweight='bold')
plt.ylabel('f(x)', color='red', fontsize=14, fontweight='bold', fontstyle='italic')
````

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlim.html

### Márgenes entre los ejes.

El método ```plt.margins()``` permite establecer márgenes entre los ejes y los datos del gráfico. Este método es útil para evitar que los datos se muestren demasiado cerca de los bordes de la figura, lo que puede dificultar su visualización.

```
plt.margins(x=<margen_x>, y=<margen_y>, **kwargs)
```
Donde:
* `<margen_x>` es el margen que se desea establecer en el eje $x$, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `0.05`, lo que significa que se establecerá un margen del 5% del rango de los datos en el eje $x$.
* `<margen_y>` es el margen que se desea establecer en el eje $y$, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `0.05`, lo que significa que se establecerá un margen del 5% del rango de los datos en el eje $y$.

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.margins.html

**Ejemplos:**

### Rejillas.

El método ```plt.grid()``` permite mostrar una rejilla en el gráfico, lo que puede facilitar la visualización de los datos y la interpretación de los valores en los ejes.

```
plt.grid(visible=<bool>, which=<tipo>, axis=<ejes>, **kwargs)
```
Donde:
* `<bool>` es un valor booleano que indica si se desea mostrar la rejilla en el gráfico. El valor por defecto es `False`, lo que significa que no se mostrará la rejilla.
* `<tipo>` es el tipo de líneas que se desea utilizar para la rejilla, el cual se puede especificar utilizando los siguientes valores:
    * `'major'` para mostrar sólo las líneas principales de la rejilla.
    * `'minor'` para mostrar sólo las líneas secundarias de la rejilla.
    * `'both'` para mostrar tanto las líneas principales como las líneas secundarias de la rejilla. El valor por defecto es `'major'`.
* `<ejes>` es el eje o los ejes en los que se desea mostrar la rejilla, el cual se puede especificar utilizando los siguientes valores:
    * `'x'` para mostrar la rejilla sólo en el eje $x$.
    * `'y'` para mostrar la rejilla sólo en el eje $y$.
    * `'both'` para mostrar la rejilla en ambos ejes. El valor por defecto es `'both'`.

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.grid.html

**Ejemplos:**

* La siguiente celda mostrará una rejilla con líneas principales y secundarias en ambos ejes utilizando el método ```plt.grid()``` con los argumentos:
   * ```visible=True```
   * ```which='both'```
   * ```axis='both'```.

````python
plt.figure(dpi=80)
plt.title("Pythonista")
plt.plot(x, y, label='función 1')
plt.plot(x, z, label='función 2')
plt.legend(loc="lower right")
plt.xlabel('Eje de las x')
plt.ylabel('f(x)')
plt.grid(visible=True, which='both', axis='both') # Mostrar la cuadrícula en el gráfico
````

* La siguiente celda mostrará una rejilla con líneas principales y secundarias en ambos ejes utilizando el método ```plt.grid()``` con los argumentos:

### Límites de los ejes y relación de aspecto.
El método ```plt.axis()``` permite establecer los límites de los ejes y la relación de aspecto del gráfico.

```
plt.axis(option=<opción>, xmin=<límite_inferior_x>, xmax=<límite_superior_x>, ymin=<límite_inferior_y>, ymax=<límite_superior_y>, **kwargs)
```
Donde:
* `<opción>` es una cadena de texto que representa la opción que se desea establecer para los ejes, la cual se puede especificar utilizando los siguientes valores:
    * `'on'` para mostrar los ejes y las etiquetas de los ejes.
    * `'off'` para ocultar los ejes y las etiquetas de los ejes.
    * `'equal'` para establecer la relación de aspecto del gráfico a 1:1, lo que significa que un incremento unitario en el eje $x$ tendrá la misma longitud que un incremento unitario en el eje $y$.
    * `'scaled'` para establecer la relación de aspecto del gráfico a 1:1, pero ajustando los límites de los ejes para que se muestren todos los datos del gráfico.
    * `'tight'` para ajustar automáticamente los límites de los ejes para que se muestren todos los datos del gráfico sin dejar espacio adicional.
    * `'auto'` para ajustar automáticamente los límites de los ejes para que se muestren todos los datos del gráfico, pero dejando un margen adicional alrededor de los datos. El valor por defecto es `'on'`.
* `<límite_inferior_x>` es el valor mínimo que se desea mostrar en el eje $x$, el cual se puede especificar como un número entero o un número de punto flotante.
* `<límite_superior_x>` es el valor máximo que se desea mostrar en el eje $x$, el cual se puede especificar como un número entero o un número de punto flotante.
* `<límite_inferior_y>` es el valor mínimo que se desea mostrar en el eje $y$, el cual se puede especificar como un número entero o un número de punto flotante.
* `<límite_superior_y>` es el valor máximo que se desea mostrar en el eje $y$, el cual se puede especificar como un número entero o un número de punto flotante.
    
https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axis.html

**Ejemplos:**

La siguiente celda trazará un gráfico con la relación de aspecto establecida a 1:1 utilizando el método ```plt.axis()``` con el argumento:
* ```'equal'```.

````python
plt.figure(dpi=80)
plt.title("Pythonista")
plt.plot(x, y, label='función 1')
plt.plot(x, z, label='función 2')
plt.legend(loc="lower right")
plt.xlabel('Eje de las x')
plt.ylabel('f(x)')
plt.axis('equal') # Establece la relación de aspecto del gráfico a 1:1
````

* La siguiente celda desplegará un gráfico con los límites de los ejes establecidos por:

## Solapamientos entre los elementos del gráfico.
En ocasiones, los elementos del gráfico pueden solaparse entre sí, lo que puede dificultar la visualización de los datos y la interpretación de los valores en los ejes. Para evitar este problema, se puede utilizar el método ```plt.tight_layout()```, el cual ajusta automáticamente el diseño de la figura para evitar solapamientos entre los elementos del gráfico.

```plt.tight_layout(pad=<margen>, w_pad=<margen_horizontal>, h_pad=<margen_vertical>, **kwargs)```

Donde:
* `<margen>` es el margen general que se desea establecer entre los elementos del gráfico, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `1.08`, lo que significa que se establecerá un margen del 8% del tamaño de la figura entre los elementos del gráfico.
* `<margen_horizontal>` es el margen horizontal que se desea establecer entre los elementos del gráfico, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `0.2`, lo que significa que se establecerá un margen del 20% del tamaño de la figura entre los elementos del gráfico en la dirección horizontal.
* `<margen_vertical>` es el margen vertical que se desea establecer entre los elementos del gráfico, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `0.2`, lo que significa que se establecerá un margen del 20% del tamaño de la figura entre los elementos del gráfico en la dirección vertical.

**Ejemplos:**

* La siguiente celda definirá un espacio entre los elementos del gráfico utilizando el método ```plt.tight_layout()``` con los argumentos:
   * ```pad=0.7```
   * ```w_pad=0.7```
   * ```h_pad=0.7```.

````python
plt.figure(dpi=80)
plt.title("Pythonista")
plt.plot(x, y, label='función 1')
plt.plot(x, z, label='función 2')
plt.legend(loc="lower right")
plt.xlabel('Eje de las x')
plt.ylabel('f(x)')
plt.tight_layout(pad=0.7, w_pad=0.7, h_pad=0.7) # Ajusta automáticamente el diseño de la figura para evitar solapamientos entre los elementos del gráfico
````

### Guardado de imagen en un archivo.
El método ```plt.savefig()``` permite guardar la figura en un archivo, lo que es útil para compartir los gráficos o para utilizarlos en informes, presentaciones, etc.
```
plt.savefig(<nombre_archivo>, dpi=<resolución>, format=<formato>, **kwargs)
```
Donde:
* `<nombre_archivo>` es una cadena de texto que representa el nombre del archivo en el que se desea guardar la figura, incluyendo la extensión del archivo (por ejemplo, `grafico.png`).
* `<resolución>` es la resolución de la imagen guardada, la cual se puede especificar como un número entero que representa los puntos por pulgada (DPI) de la imagen. El valor por defecto es `100`.
* `<formato>` es el formato de archivo en el que se desea guardar la figura, el cual se puede especificar utilizando los siguientes valores:
    * `'png'` para guardar la figura en formato PNG.
    * `'jpg'` o `'jpeg'` para guardar la figura en formato JPEG.
    * `'pdf'` para guardar la figura en formato PDF.
    * `'svg'` para guardar la figura en formato SVG.
    * `'eps'` para guardar la figura en formato EPS.
    * `'ps'` para guardar la figura en formato PS.
    * `'tiff'` para guardar la figura en formato TIFF.
    * `'bmp'` para guardar la figura en formato BMP.
    * `'raw'` para guardar la figura en formato RAW.
    * `'rgba'` para guardar la figura en formato RGBA.
    * `'jpg'` para guardar la figura en formato JPEG.
    * `'jpeg'` para guardar la figura en formato JPEG.

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.savefig.html

Donde:

* Cuadernos de Jupyter.

El método `plt.show()` es compatible con todos estos entornos y permite mostrar los gráficos de forma interactiva, lo que facilita la visualización de los datos y la interpretación de los valores en los ejes.

### Cierre de la figura.
El método ```plt.close()``` permite cerrar la figura, lo que es útil para liberar recursos y evitar que se muestren figuras no deseadas.

Este método se puede utilizar de las siguientes formas:
* ```plt.close()``` para cerrar la figura actual.
* ```plt.close(fig)``` para cerrar una figura específica, donde `fig` es el objeto *figure* que se desea cerrar.

 En el caso de *IPython/Jupyter*, el método `plt.close()` es especialmente útil para cerrar las figuras que se han mostrado en el cuaderno, lo que puede ayudar a mantener el cuaderno organizado y evitar la acumulación de figuras no deseadas.

## Estilos.
Los estilos en *Matplotlib* permiten personalizar la apariencia de los gráficos de forma rápida y sencilla, sin necesidad de configurar manualmente cada uno de los elementos del gráfico. *Matplotlib* ofrece una variedad de estilos predefinidos que se pueden utilizar para cambiar el aspecto de los gráficos con un solo comando.

Para utilizar un estilo predefinido en *Matplotlib*, se puede utilizar el método `plt.style.use()`, el cual permite aplicar un estilo específico a todos los gráficos que se tracen después de llamar a este método.

Una vez que se ha aplicado un estilo utilizando `plt.style.use()`, todos los gráficos que se tracen a continuación utilizarán ese estilo de forma predeterminada, lo que facilita la creación de gráficos con una apariencia consistente y atractiva.

```
plt.style.use(<estilo>)
```
Donde:
* `<estilo>` es una cadena de texto que representa el nombre del estilo que se desea aplicar a los gráficos, el cual se puede especificar utilizando los siguientes valores:
    * `'default'` para aplicar el estilo predeterminado de *Matplotlib*.
    * `'classic'` para aplicar el estilo clásico de *Matplotlib*.
    * `'ggplot'` para aplicar el estilo inspirado en la biblioteca *ggplot2* de R.
    * `'seaborn'` para aplicar el estilo inspirado en la biblioteca *Seaborn* de Python.
    * `'bmh'` para aplicar el estilo inspirado en la biblioteca *Bayesian Methods for Hackers*.
    * `'dark_background'` para aplicar un estilo con fondo oscuro.
    * `'fivethirtyeight'` para aplicar el estilo inspirado en el sitio web *FiveThirtyEight*.
    * `'grayscale'` para aplicar un estilo en escala de grises.
    * `'Solarize_Light2'` para aplicar un estilo inspirado en el tema Solarized Light.
    * `'tableau-colorblind10'` para aplicar un estilo inspirado en la paleta de colores para daltónicos de Tableau.
    * `'seaborn-darkgrid'` para aplicar un estilo inspirado en la biblioteca *Seaborn* con fondo oscuro y rejilla.
    * `'seaborn-notebook'` para aplicar un estilo inspirado en la biblioteca *Seaborn* con un diseño optimizado para cuadernos de Jupyter.
    * `'seaborn-poster'` para aplicar un estilo inspirado en la biblioteca *Seaborn* con un diseño optimizado para presentaciones.

https://matplotlib.org/stable/gallery/style_sheets/style_sheets_reference.html

## Imágenes con *Matplotlib*.
*Matplotlib* también permite trabajar con imágenes, lo que es útil para visualizar datos en formato de imagen, como por ejemplo:
* Imágenes en formato PNG, JPEG, etc.
* Imágenes en formato TIFF, BMP, etc.
* Imágenes en formato RAW, RGBA, etc.
Para trabajar con imágenes en *Matplotlib*, se pueden utilizar los siguientes métodos:
* [`plt.imread()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imread.html) para leer una imagen desde un archivo y cargarla en un arreglo de NumPy.
* [`plt.imshow()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html) para mostrar una imagen en un gráfico.
* [`plt.imsave()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imsave.html) para guardar una imagen en un archivo.

### Carga de imágenes.
El método `plt.imread()` permite leer una imagen desde un archivo y cargarla en un arreglo de *NumPy*, lo que facilita su manipulación y visualización en *Matplotlib*.

```
plt.imread(<nombre_archivo>, format=<formato>, **kwargs)
```
Donde:
* `<nombre_archivo>` es una cadena de texto que representa el nombre del archivo de imagen que se desea cargar, incluyendo la extensión del archivo (por ejemplo, `imagen.png`).
* `<formato>` es el formato de la imagen que se desea cargar, el cual se puede especificar utilizando los siguientes valores:
    * `'png'` para cargar una imagen en formato PNG.
    * `'jpg'` o `'jpeg'` para cargar una imagen en formato JPEG.
    * `'tiff'` para cargar una imagen en formato TIFF.
    * `'bmp'` para cargar una imagen en formato BMP.
    * `'raw'` para cargar una imagen en formato RAW.
    * `'rgba'` para cargar una imagen en formato RGBA.

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imread.html

* La siguiente celda cargará una imagen desde un archivo llamado `img/pythonista.png`.

````python
imagen = plt.imread('img/pythonista.png')
````

* El objeto `imagen` es un arreglo de *NumPy* que contiene los datos de la imagen cargada desde el archivo `img/pythonista.png`. Este arreglo tiene una forma de `(alto, ancho, canales)`, donde `alto` es el número de píxeles en la dirección vertical, `ancho` es el número de píxeles en la dirección horizontal, y `canales` es el número de canales de color en la imagen (por ejemplo, 3 para imágenes RGB y 4 para imágenes RGBA). Cada elemento del arreglo representa el valor de intensidad de un píxel en la imagen, con valores que van desde `0` hasta `255` para imágenes en formato `int8`.

### Despliegue de imágenes.
El método `plt.imshow()` permite mostrar una imagen en un gráfico, lo que es útil para visualizar datos en formato de imagen y para crear gráficos que combinan elementos gráficos con imágenes.
```
plt.imshow(<imagen>, cmap=<colormap>, vmin=<valor_mínimo>, vmax=<valor_máximo>, **kwargs)
```
Donde:
* `<imagen>` es un arreglo de NumPy que contiene los datos de la imagen que se desea mostrar en el gráfico. Este arreglo debe tener una forma de `(alto, ancho)` para imágenes en escala de grises, o una forma de `(alto, ancho, canales)` para imágenes en formato RGB o RGBA.
* `<colormap>` es el mapa de colores que se desea utilizar para mostrar la imagen, el cual se puede especificar utilizando los siguientes valores:
    * `'viridis'` para utilizar el mapa de colores Viridis.
    * `'plasma'` para utilizar el mapa de colores Plasma.
    * `'inferno'` para utilizar el mapa de colores Inferno.
    * `'magma'` para utilizar el mapa de colores Magma.
    * `'cividis'` para utilizar el mapa de colores Cividis.
    * `'gray'` para utilizar el mapa de colores en escala de grises.
    * `'hot'` para utilizar el mapa de colores Hot.
    * `'cool'` para utilizar el mapa de colores Cool.
    * `'spring'` para utilizar el mapa de colores Spring.
    * `'summer'` para utilizar el mapa de colores Summer.
    * `'autumn'` para utilizar el mapa de colores Autumn.
    * `'winter'` para utilizar el mapa de colores Winter.
* `<valor_mínimo>` es el valor mínimo que se desea mostrar en la imagen, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `None`, lo que significa que se utilizará el valor mínimo del arreglo de la imagen.
* `<valor_máximo>` es el valor máximo que se desea mostrar en la imagen, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `None`, lo que significa que se utilizará el valor máximo del arreglo de la imagen.

* https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html

* La siguiente celda desplegará la imagen cargada en el arreglo `imagen` utilizando el método `plt.imshow()` con el argumento:
   * `cmap='viridis'` para mostrar la imagen utilizando el mapa de colores Viridis.

````python
plt.imshow(imagen, cmap='viridis') # Mostrar la imagen utilizando el mapa de colores 'viridis', que es un mapa de colores perceptualmente uniforme que va desde el azul oscuro hasta el amarillo brillante, ideal para representar datos con una amplia gama de valores.
plt.axis('off') # Ocultar los ejes para mostrar solo la imagen
````

* `<valor_mínimo>` es el valor mínimo que se desea mostrar en la imagen, el cual se puede especificar como un número entero o un número de punto flotante. El valor por defecto es `None`, lo que significa que se utilizará el valor mínimo del arreglo de la imagen.
