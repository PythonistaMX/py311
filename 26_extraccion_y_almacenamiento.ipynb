## Obtención de datos abiertos del Banco Mundial.

El Banco Mundial publica en *World Bank Open Data* miles de indicadores económicos y sociales de casi todos los países, con la licencia *Creative Commons* Atribución 4.0 (CC BY 4.0), que permite usarlos y redistribuirlos citando la fuente.

En este capítulo se usará el indicador `NY.GDP.MKTP.KD`: el producto interno bruto (PIB) anual de cada país, en dólares estadounidenses a **precios constantes de 2015**. El PIB mide el valor de todos los bienes y servicios que produce un país en un año. Los precios constantes eliminan el efecto de la inflación, de modo que el cambio de un año a otro refleja el cambio en el volumen de la producción.

La página del indicador se encuentra en:

https://data.worldbank.org/indicator/NY.GDP.MKTP.KD

### La API del Banco Mundial.

Una API (*Application Programming Interface*, interfaz de programación de aplicaciones) web es un servicio que responde a una dirección URL con datos en lugar de una página para personas. La API del Banco Mundial no requiere registro ni claves de acceso y puede responder en formato JSON (*JavaScript Object Notation*), un formato de texto para datos estructurados que combina objetos, equivalentes a los diccionarios de *Python*, y listas.

La dirección de una consulta de un indicador tiene la siguiente forma:

```
https://api.worldbank.org/v2/<idioma>/country/<países>/indicator/<indicador>?format=json&date=<inicio>:<fin>&per_page=<n>
```

Donde:

* `<idioma>` es el idioma de los nombres de países y regiones, por ejemplo `es` para español o `en` para inglés.
* `<países>` son los códigos de tres letras de los países según la norma ISO 3166-1 alfa-3 de la Organización Internacional de Normalización (ISO), separados por `;`. El código `WLD` corresponde al agregado mundial.
* `<indicador>` es el código del indicador.
* `<inicio>:<fin>` es el rango de años.
* `<n>` es el número máximo de registros por página de la respuesta.

La documentación de la API se encuentra en:

https://datahelpdesk.worldbank.org/knowledgebase/articles/889392-about-the-indicators-api-documentation

**ADVERTENCIA:** El Banco Mundial revisa sus cifras periódicamente, así que una descarga posterior puede tener valores distintos a los de este capítulo. El capítulo 27 usa una copia fija de estos datos, consultada el 27 de noviembre de 2026.

### Conversión a *dataframe* con `pd.json_normalize()`.

La función `pd.json_normalize()` convierte una lista de diccionarios en un *dataframe*. Las claves de los diccionarios anidados se convierten en columnas propias.

```
pd.json_normalize(<datos>, sep=<separador>)
```

Donde:

* `<datos>` es un diccionario o una lista de diccionarios.
* `<separador>` es el texto que une los nombres de las claves anidadas. Por omisión es `'.'`, de modo que la clave `'value'` dentro de `'country'` se convierte en la columna `'country.value'`.

La documentación de `pd.json_normalize()` se encuentra en: https://pandas.pydata.org/docs/reference/api/pandas.json_normalize.html

* La siguiente celda creará el *dataframe* `original` a partir de `registros` con `pd.json_normalize()`.

````python
original = pd.json_normalize(registros)
original
````

El *dataframe* `original` tiene un renglón por país y año (formato largo) y más columnas de las necesarias:

* `countryiso3code` y `country.value` contienen el código y el nombre del país.
* `date` contiene el año y `value` el valor del PIB.
* `indicator.id`, `indicator.value`, `unit`, `obs_status` y `decimal` describen el indicador y son iguales en todos los renglones.

### Selección y limpieza de columnas.

## Operaciones con archivos de formato *CSV*.

El formato *CSV* (*Comma-Separated Values*) es uno de los más utilizados para el intercambio de datos tabulares por su simplicidad y compatibilidad universal. 

*Pandas* permite exportar dataframes con el método `df.to_csv()` e importarlos con `pd.read_csv()`.

Aún cuando el formato *CSV* es ampliamente compatible, tiene limitaciones como la falta de soporte para tipos de datos complejos, problemas con caracteres especiales y la ausencia de metadatos. Por ello, es importante considerar estas limitaciones al elegir el formato de almacenamiento para los datos. 

Un archivo *CSV* sólo contiene texto: no guarda el tipo de cada columna ni la estructura del índice. Al leerlo, `pd.read_csv()` infiere los tipos a partir del contenido, y los encabezados siempre se leen como cadenas de caracteres.

* La documentación de `pd.read_csv()` está disponible en:
https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html

* La documentación de `df.to_csv()` está disponible en:
https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_csv.html

* La siguiente celda creará la variable `ruta_csv` con la ruta local del archivo `data_tmp/pib_banco_mundial.csv`, utilizando `os.path.join()`.

````python
ruta_csv = os.path.join("data_tmp", "pib_banco_mundial.csv")
````

* La siguiente celda exportará el *dataframe* `pib` en formato *CSV* con el método `to_csv()` a la ruta `ruta_csv`.
    * El método `reset_index()` convierte los niveles `codigo` y `pais` del índice en columnas ordinarias antes de exportar.
    * El argumento `index=False` evita escribir el índice numérico que genera `reset_index()`.
    * El argumento `encoding='utf-8'` especifica la codificación de caracteres del archivo, para que los acentos se guarden correctamente.

````python
pib.reset_index().to_csv(ruta_csv, index=False, encoding='utf-8')
````

* La siguiente celda leerá el archivo *CSV* recién creado con `pd.read_csv()` y lo guardará en el *dataframe* `nuevos_datos`.
   * El argumento `header=0` indica que la primera fila del archivo contiene los nombres de las columnas.
   * El argumento `encoding='utf-8'` especifica la codificación de caracteres del archivo al leerlo.

````python
nuevos_datos = pd.read_csv(ruta_csv, header=0, encoding='utf-8')
nuevos_datos
````

* La siguiente celda mostrará que `nuevos_datos` ya no tiene el `MultiIndex`: `codigo` y `pais` son columnas ordinarias y el índice es numérico.

````python
nuevos_datos.index
````

* La siguiente celda mostrará que los años, que en `pib` eran números enteros, regresaron como cadenas de caracteres: la etiqueta `'2005'` existe en `nuevos_datos`, pero la etiqueta `2005` no.

````python
'2005' in nuevos_datos.columns, 2005 in nuevos_datos.columns
````

## Operaciones con archivos de formato *parquet*.

El formato *Parquet* es un formato de almacenamiento columnar optimizado para el rendimiento y la compresión, ampliamente utilizado en ecosistemas de big data. *Pandas* permite exportar dataframes a *Parquet* con el método `df.to_parquet()` e importarlos con `pd.read_parquet()`.

* La documentación de `pd.read_parquet()` está disponible en: https://pandas.pydata.org/docs/reference/api/pandas.read_parquet.html
* La documentación de `df.to_parquet()` está disponible en: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_parquet.html

* La siguiente celda definirá la variable `ruta_parquet` con la ruta local del archivo `data_tmp/pib_banco_mundial.parquet`, utilizando `os.path.join()`.

````python
ruta_parquet = os.path.join("data_tmp", "pib_banco_mundial.parquet")
````

* La siguiente celda guardará el *dataframe* `pib` en formato *Parquet* en la ruta `ruta_parquet`.
   * El argumento `index=True` incluye el índice del *dataframe* en el archivo.
   * El argumento `engine='pyarrow'` indica el motor de escritura, en este caso la biblioteca `pyarrow`, que debe estar instalada y que se estudiará en capítulos posteriores.

````python
pib.to_parquet(ruta_parquet, index=True, engine='pyarrow')
````

* La siguiente celda cargará el archivo recién creado con `pd.read_parquet()` y lo guardará en el *dataframe* `datos_parquet`.

````python
datos_parquet = pd.read_parquet(ruta_parquet, engine='pyarrow')
datos_parquet
````

* La siguiente celda mostrará que `datos_parquet` recupera el `MultiIndex` de los renglones y que los años de las columnas siguen siendo números enteros, a diferencia del formato *CSV*. Al escribir, *PyArrow* guarda en el archivo el tipo de cada columna y los metadatos del índice y de las columnas, de modo que la lectura reconstruye el *dataframe* sin conversiones adicionales.

````python
datos_parquet.index, datos_parquet.columns[:3]
````

En el capítulo 27 se utilizará una copia fija del mismo indicador, con más países, guardada en `data/pib_banco_mundial.csv`. Así, los ejemplos de visualización no dependen de fuentes externas (disponibilidad de la *URL*, cambios de formato o revisiones de las cifras) y siempre producen el mismo resultado.
